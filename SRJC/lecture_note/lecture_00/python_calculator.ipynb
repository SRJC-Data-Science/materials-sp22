### Assignment Submission ###

#### Instructions ####

Follow the instructions below to complete this assignment:

1. Complete the assignment.
2. In the Jupyter Notebook window, go to File, then Download. This will download the .ipynb file from the cloud to your desktop.
3. In the same Notebook window, go to File, select "Save and Export Notebook As", then select "HTML". This will export your notebook to an HTML file on your desktop.
4. Submit both files to this assignment in Canvas.
